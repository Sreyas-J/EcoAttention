```python
import struct

def hex_bfloat16_to_decimal(hex_str):
    """
    Converts a hexadecimal string (bfloat16 representation) to a decimal float.
    """
    # 1. Clean the input (remove '0x' if present)
    clean_hex = hex_str.lower().replace('0x', '')
    
    # Check if length is valid for 16-bit
    if len(clean_hex) > 4:
        raise ValueError("Input hex is too long for 16-bit bfloat16 (max 4 chars).")

    # 2. Convert hex string to integer
    int16_val = int(clean_hex, 16)

    # 3. Shift left by 16 bits.
    #    bfloat16 is the first 16 bits of float32. 
    #    We pad the last 16 bits with zeros to recreate a valid float32 pattern.
    int32_val = int16_val << 16

    # 4. Pack as unsigned int (I) and unpack as float (f)
    #    '>' denotes big-endian to ensure the shift aligns correctly in memory.
    packed_bytes = struct.pack('>I', int32_val)
    decimal_val = struct.unpack('>f', packed_bytes)[0]

    return decimal_val

# --- Test Cases ---
# tests = [
#     "3f80",  # 1.0
#     "4049",  # ~3.14 (Pi approximation)
#     "c000",  # -2.0
#     "0000",  # 0.0
#     "8000",  # -0.0
#     "7f80",  # Infinity
#     "ff80",  # -Infinity
# ]

# print(f"{'Hex':<10} | {'Decimal':<20}")
# print("-" * 30)

# for h in tests:
#     val = hex_bfloat16_to_decimal(h)
#     print(f"0x{h:<8} | {val:<20}")
print(hex_bfloat16_to_decimal("0800"))
```
